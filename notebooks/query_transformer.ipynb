```python
from langchain_core.prompts import PromptTemplate
from langchain.chat_models import init_chat_model
from langchain_groq import ChatGroq
```

```python
class query_transform:
    def __init__(self):
        self.model=ChatGroq("groq:openai/gpt-oss-120b",temprature=0.2,max_tokens=4000)
        """ 
                rewrite the orginal query to improve the retrival
                Args:
                orginal_query(str): the orginal query
                Returns:
                str: the rewriten query
        """
            
    def rewrite_query(self,orginal_query):
        try:
            query_rewrite_template=""" you are AI assistant tasked with reformulating user querys to improve retrival in a RAG system. 
            Given The orginal_query rewrite it to be more specific and likely to be retrive relevant information.
            Orginal_query:{'orginal_query'}
            Rewriten query
            """
            query_rewrite_prompt=PromptTemplate(input_variables=['orginal_query'],
                                                template=query_rewrite_template)
            query_rewriter=query_rewrite_prompt|self.model
            response=query_rewriter.invoke(orginal_query)
            return response.content
        except Exception:
            logging.error('enter the query')


        
    
```
