```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
import logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s: %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)

logger = logging.getLogger()

```

```python

class chunking:
    def __init__(self,document=None):
        self.chunk=None
        self.document=document or []
        self.overlap=None
        self.size=None
    def total_character(self,document):
        return sum(len(doc.page_content) for doc in document)
    def pick_size(self):
        n=self.total_character()
        try:
             if n < 5_000:
                 size = 300
             elif n < 50_000:         
                 size = 500
             elif n < 500_000:        
                 size = 1000
             else:                    
                 size = 1500
             self.size=size
             self.overlap=int(size*0.15)
             return self.size,self.overlap
        except Exception as e:
            logging.exception('unable to get the size of the doc')
    def text_split(self,chunk_size=None,chunk_overlap=None):
        if not self.document:
            logging.warning('no document to chunk')
            self.chunk=[]
            return self.chunk
        if chunk_size is None or chunk_overlap is None:
            self.pick_size()
            chunk_size=chunk_size or self.size
            chunk_overlap=chunk_overlap or self.overlap
        splitter= RecursiveCharacterTextSplitter(chunk_size=chunk_size, chunk_overlap=chunk_overlap,separators=["\n\n", "\n", ".", " ", ""])
        self.chunk=splitter.split_documents(self.document)
        logger.info(f'document chunked into {len(self.chunk)} chunks with size{chunk_size} and overlap {chunk_overlap}')
        return self.chunk

```
